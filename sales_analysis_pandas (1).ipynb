# **Python code**

```python
import pandas as pd

df = pd.read_csv("/sales.csv.txt")

df
```

# **Data Basic Information**

```python
df.info()
```

# **Revenue**

```python
df["Revenue"] = df["Quantity"] * df["Price"]

df
```

# **Total Revenue**

```python
total_revenue = df["Revenue"].sum()

print("Total Revenue:", total_revenue)
```

# **Product-Wise Analysis**

```python
product_summary = df.groupby("Product").agg(
    Total_Quantity=("Quantity", "sum"),
    Total_Revenue=("Revenue", "sum")
).reset_index()

product_summary
```

# **Top-Selling Product**

```python
top_product = product_summary.loc[
    product_summary["Total_Quantity"].idxmax()
]

print("Top-selling product:", top_product["Product"])
print("Units sold:", top_product["Total_Quantity"])
```

# **Highest Revenue Product**

```python
highest_revenue_product = product_summary.loc[
    product_summary["Total_Revenue"].idxmax()
]

print("Highest revenue product:", highest_revenue_product["Product"])
print("Revenue:", highest_revenue_product["Total_Revenue"])
```

# **Summary**

```python
print("===== SALES DATA ANALYSIS =====")
print("Total Revenue: ₹", total_revenue)
print("Top-selling Product:", top_product["Product"])
print("Units Sold:", top_product["Total_Quantity"])
print("Highest Revenue Product:", highest_revenue_product["Product"])
print("Highest Product Revenue: ₹", highest_revenue_product["Total_Revenue"])
```
